```python
#import libraries
import spacy

# Load English NLP model
nlp = spacy.load("en_core_web_sm")
```

```python
#import paragraphy
text = """
John is a student. He studies computer science.
Mary is his friend. She helps him with his projects.
They often work together because they want to improve their skills.
"""

doc = nlp(text)

print("INPUT PARAGRAPH:")
print(text)
```

```python
#Discourse Structure Analysis
print("\nDISCOURSE STRUCTURE ANALYSIS")

# Sentence segmentation
print("\nSentences:")
for i, sent in enumerate(doc.sents, 1):
    print(f"Sentence {i}: {sent.text}")

# Identify noun phrases
print("\nNoun Phrases:")
for chunk in doc.noun_chunks:
    print(chunk.text)

# Identify verbs
print("\nVerbs:")
for token in doc:
    if token.pos_ == "VERB":
        print(token.text)

# Identify important entities
print("\nNamed Entities:")
for ent in doc.ents:
    print(ent.text, "-->", ent.label_)
```

```python
#Reference Resolution
print("\nREFERENCE RESOLUTION")

# Store noun phrases
noun_phrases = list(doc.noun_chunks)

# Identify pronouns
pronouns = [
    token for token in doc
    if token.pos_ == "PRON"
]

# Simple reference resolution
for pronoun in pronouns:

    candidates = []

    for np in noun_phrases:

        # Noun phrase must appear before pronoun
        if np.end <= pronoun.i:

            # Check gender/number compatibility
            if pronoun.text.lower() in ["he", "him", "his"]:
                if any(t.pos_ == "PROPN" for t in np):
                    candidates.append(np)

            elif pronoun.text.lower() in ["she", "her"]:
                if any(t.pos_ == "PROPN" for t in np):
                    candidates.append(np)

            elif pronoun.text.lower() in ["they", "them", "their"]:
                candidates.append(np)

            else:
                candidates.append(np)

    if candidates:
        reference = candidates[-1]
        print(f"Pronoun: {pronoun.text} --> Referent: {reference.text}")
    else:
        print(f"Pronoun: {pronoun.text} --> Referent: Not found")
```

```python
#Discourse Coherence Analysis
print("\nDISCOURSE COHERENCE ANALYSIS")

sentences = list(doc.sents)

# 1. Count sentences
print("Total Sentences:", len(sentences))

# 2. Identify conjunctions
print("\nConnecting Words:")

connectors = []

for token in doc:
    if token.pos_ == "CCONJ" or token.pos_ == "SCONJ":
        connectors.append(token.text)

print(connectors)

# 3. Identify repeated important words
word_frequency = {}

for token in doc:
    if token.is_alpha and not token.is_stop:
        word = token.lemma_.lower()
        word_frequency[word] = word_frequency.get(word, 0) + 1

print("\nRepeated Important Words:")

for word, count in word_frequency.items():
    if count > 1:
        print(word, ":", count)

# 4. Analyze sentence connectivity
if len(connectors) > 0:
    print("\nCoherence: Connecting words are present.")
else:
    print("\nCoherence: Few explicit connecting words found.")

# 5. Display sentence structure
print("\nSentence Connectivity:")

for i in range(len(sentences) - 1):
    print(
        f"Sentence {i+1} -> Sentence {i+2}"
    )
```
