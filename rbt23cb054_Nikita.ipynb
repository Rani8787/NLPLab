```python
import spacy


nlp = spacy.load("en_core_web_sm")

text ="I love NLP. NLP is a easy subject"
text = """Rahul bought a new laptop. He uses it every day.
The laptop is very useful. It helps him complete his work."""

doc = nlp(text)

print("Discourse Structure:")
for sent in doc.sents:
    print("-", sent.text)

print("\nPronoun Reference Resolution:")

for token in doc:
    if token.pos_ == "PRON":
        print(token.text, "->", "Possible noun reference")

print("\nDiscourse Coherence:")

sentences = list(doc.sents)

for i in range(len(sentences) - 1):
    words1 = set(token.lemma_.lower() for token in sentences[i]
                 if token.is_alpha)
    words2 = set(token.lemma_.lower() for token in sentences[i+1]
                 if token.is_alpha)

    common = words1.intersection(words2)

    print(f"Sentence {i+1} and {i+2}:")
    print("Common words:", common)
    print("Coherence:", "Good" if common else "Weak")
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = """Rahul bought a new laptop. He uses it every day.
The laptop is very useful. It helps him complete his work."""

doc = nlp(text)

# 1. Discourse Structure
print("Discourse Structure:")
sentences = list(doc.sents)

for sent in sentences:
    print("-", sent.text)

# 2. Pronoun Detection
print("\nPronoun Reference Resolution:")

for token in doc:
    if token.pos_ == "PRON":
        print(token.text)

# 3. Discourse Coherence
print("\nDiscourse Coherence:")

for i in range(len(sentences) - 1):

    words1 = set(t.lemma_.lower() for t in sentences[i]
                 if t.is_alpha)

    words2 = set(t.lemma_.lower() for t in sentences[i+1]
                 if t.is_alpha)

    pronouns = {"he", "she", "it", "they", "him",
                "her", "them", "his", "their"}

    common = words1.intersection(words2)

    if common or any(t.text.lower() in pronouns
                     for t in sentences[i+1]):
        result = "Connected"
    else:
        result = "Not Connected"

    print(f"Sentence {i+1} and {i+2}: {result}")
```
