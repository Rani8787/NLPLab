```python
# Pronoun Resolution and Coreference System

!pip install -q spacy
!python -m spacy download en_core_web_sm -q

import spacy

# Load English NLP model
nlp = spacy.load("en_core_web_sm")

# Sample paragraph
text = """
Riya went to the market with her brother.
She bought a book for him.
The book was expensive, but he liked it.
"""

# Process paragraph
doc = nlp(text)

# Extract noun phrases
noun_phrases = list(doc.noun_chunks)

# Person-related words
person_words = {
    "boy", "girl", "man", "woman", "brother",
    "sister", "father", "mother", "teacher",
    "student", "person", "friend"
}

# Identify person noun phrases
def is_person(chunk):
    root = chunk.root
    return (
        root.pos_ == "PROPN"
        or root.lemma_.lower() in person_words
    )

# Pronoun resolution
def resolve_pronoun(pronoun, candidates):
    p = pronoun.text.lower()

    if p in ["he", "him", "his", "she", "her", "hers"]:
        valid = [c for c in candidates if is_person(c)]

    elif p in ["it", "its"]:
        valid = [c for c in candidates if not is_person(c)]

    elif p in ["they", "them", "their", "theirs"]:
        valid = [
            c for c in candidates
            if "Plur" in c.root.morph.get("Number")
        ]

    else:
        valid = candidates

    # Select nearest previous matching noun phrase
    if valid:
        return valid[-1].text

    if candidates:
        return candidates[-1].text

    return "No reference found"

# Find pronouns and their references
print("PRONOUN RESOLUTION AND COREFERENCE")
print("----------------------------------")

for token in doc:
    if token.pos_ == "PRON":
        previous_nouns = [
            chunk for chunk in noun_phrases
            if chunk.end <= token.i
        ]

        reference = resolve_pronoun(token, previous_nouns)

        print(f"Pronoun: {token.text:10} --> Reference: {reference}")

# Display noun phrases
print("\nIdentified Noun Phrases:")
for chunk in noun_phrases:
    print("-", chunk.text)
```
