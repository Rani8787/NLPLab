```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')
```

```python
text = """
Natural Language Processing (NLP) is an exciting field of Artificial Intelligence.
Students are learning NLP in the laboratory.
It helps computers understand human language efficiently.
"""

print("="*80)
print("ORIGINAL TEXT")
print("="*80)
print(text)
```

```python
tokens = word_tokenize(text)

print("\n" + "="*80)
print("1. TOKENIZATION")
print("="*80)
print(tokens)
```

```python
stop_words = set(stopwords.words("english"))

filtered_words = [word for word in tokens if word.lower() not in stop_words]

print("\n" + "="*80)
print("2. STOP WORD REMOVAL")
print("="*80)
print(filtered_words)
```

```python
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print("\n" + "="*80)
print("3. STEMMING")
print("="*80)
print(stemmed_words)
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("\n" + "="*80)
print("4. LEMMATIZATION")
print("="*80)
print(lemmatized_words)
```

```python
print("\n" + "="*80)
print("COMPARISON TABLE")
print("="*80)

print("{:<20} {:<20} {:<20}".format("Original","Stemmed","Lemmatized"))
print("-"*80)

for o,s,l in zip(filtered_words, stemmed_words, lemmatized_words):
    print("{:<20} {:<20} {:<20}".format(o,s,l))
```
